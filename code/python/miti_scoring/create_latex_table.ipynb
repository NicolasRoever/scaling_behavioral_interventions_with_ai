```python
df_global = pd.read_csv("output/validation_global_scores_results.csv")
df_behavioral = pd.read_csv("output/behavioral_scores_validation_results.csv")
```

```python
def make_latex_table(df_global, df_behavioral, out_path="table_skeleton.tex"):
    def row_to_latex(name, bias, corr):
        # round correlation to 2 decimals, allow missing values
        corr_str = "" if pd.isna(corr) else f"{corr:.2f}"
        bias_str = "" if pd.isna(bias) else f"{bias:.2f}"
        return f"{name} & {bias_str} & {corr_str} \\\\"

    lines = []
    lines.append(r"\begin{tabular}{p{8cm} c c}")
    lines.append(r"\toprule")
    lines.append(r"\textbf{Score Category} & \textbf{Bias} & \textbf{Correlation} \\")
    
    # A. Global Metrics
    lines.append(r"\textit{A. Global Metrics} \\")
    for _, row in df_global.iterrows():
        lines.append(row_to_latex(row['Dimension'], row['Bias'], row['Correlation']))
    
    lines.append(r"\addlinespace")
    # B. Behavioral Counts
    lines.append(r"\textit{B. Behavioral Counts} \\")
    for _, row in df_behavioral.iterrows():
        lines.append(row_to_latex(row['Category'], row['Bias'], row['Correlation']))
    
    lines.append(r"\bottomrule")
    lines.append(r"\end{tabular}")

    # Write to .tex file
    with open(out_path, "w") as f:
        f.write("\n".join(lines))

    print(f"LaTeX skeleton saved to {out_path}")

```

```python
make_latex_table(df_global, df_behavioral, "/path/to/project/code/analysis_NR/6731ca401220dcd3b28dc2ec/tables/mi_validation_results_table.tex")
```
