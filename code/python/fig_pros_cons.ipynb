```python
from helper import *
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("output/df_clean_with_llm_themes_pros_cons_v001.csv")
```

```python
df[['number_positive_statements', 'number_negative_statements']] = df['llm_posiitives_negatives_openai'].str.extract(
    r"'positive'\s*:\s*([0-9]+).*?'negative'\s*:\s*([0-9]+)"
).apply(pd.to_numeric)
df["T"] = df["T"].replace("Ambivalence", "Decisional Balance")
```

```python
fig, axes = plt.subplots(2, 1, figsize=(9, 8), sharex=False)

# Positive statements
sns.kdeplot(
    data=df,
    x="number_positive_statements",
    hue="T",
    fill=True,
    common_norm=False,
    alpha=0.35,
    ax=axes[0]
)
axes[0].set_title("Density of number of positive statements by T")
axes[0].set_xlabel("Number of positive statements")
axes[0].set_ylabel("Density")

# Negative statements
sns.kdeplot(
    data=df,
    x="number_negative_statements",
    hue="T",
    fill=True,
    common_norm=False,
    alpha=0.35,
    ax=axes[1]
)
axes[1].set_title("Density of number of negative statements by T")
axes[1].set_xlabel("Number of negative statements")
axes[1].set_ylabel("Density")

plt.tight_layout()
plt.show()
```
